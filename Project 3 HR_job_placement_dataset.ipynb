```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from reportlab.lib import colors
from reportlab.lib.pagesizes import letter
from reportlab.lib.styles import ParagraphStyle, getSampleStyleSheet
from reportlab.platypus import (
    HRFlowable,
    Image,
    PageBreak,
    Paragraph,
    SimpleDocTemplate,
    Spacer,
    Table,
    TableStyle,
)


def generate_automated_report(df, output_pdf="Placement_Analysis_Report.pdf"):
  # =====================================================
  # 1. CALCULATE AUTOMATED INSIGHTS
  # =====================================================
  total_candidates = len(df)
  placed_count = len(df[df["status"] == "Placed"])
  not_placed_count = len(df[df["status"] == "Not Placed"])
  placement_rate = (
      (placed_count / total_candidates) * 100 if total_candidates > 0 else 0
  )

  avg_tech = df["technical_score"].mean() if "technical_score" in df else 0
  avg_skill = (
      df["skills_match_percentage"].mean()
      if "skills_match_percentage" in df
      else 0
  )
  avg_exp = (
      df["years_of_experience"].mean() if "years_of_experience" in df else 0
  )

  # Automated conclusion text based on data threshold
  if placement_rate >= 60:
    performance_verdict = (
        "Strong overall placement performance driven by solid technical scores"
        " and skills match alignment."
    )
    rec_text = "Maintain current training programs while scaling corporate hiring partnerships."
  else:
    performance_verdict = (
        "Suboptimal placement rates indicating gaps in skill readiness or expected"
        " CTC misalignments."
    )
    rec_text = "Focus on targeted technical upskilling and interview preparation bootcamps."

  # =====================================================
  # 2. GENERATE AUTOMATED CHARTS
  # =====================================================
  pie_chart_path = "temp_pie_chart.png"
  plt.figure(figsize=(4, 3))
  df["status"].value_counts().plot(
      kind="pie",
      autopct="%1.1f%%",
      colors=["#2b5c8f", "#d95f02"],
      startangle=90,
  )
  plt.title("Placement Status Breakdown", fontsize=10, fontweight="bold")
  plt.ylabel("")
  plt.tight_layout()
  plt.savefig(pie_chart_path, dpi=200)
  plt.close()

  # =====================================================
  # 3. BUILD PDF WITH REPORTLAB
  # =====================================================
  doc = SimpleDocTemplate(
      output_pdf,
      pagesize=letter,
      rightMargin=36,
      leftMargin=36,
      topMargin=36,
      bottomMargin=36,
  )
  styles = getSampleStyleSheet()

  # Custom Styles
  title_style = ParagraphStyle(
      "DocTitle",
      parent=styles["Heading1"],
      fontSize=22,
      textColor=colors.HexColor("#1A365D"),
      spaceAfter=6,
  )
  subtitle_style = ParagraphStyle(
      "SubTitle",
      parent=styles["Normal"],
      fontSize=11,
      textColor=colors.HexColor("#4A5568"),
      spaceAfter=15,
  )
  h2_style = ParagraphStyle(
      "Heading2",
      parent=styles["Heading2"],
      fontSize=14,
      textColor=colors.HexColor("#2B6CB0"),
      spaceBefore=12,
      spaceAfter=8,
  )
  body_style = ParagraphStyle(
      "Body",
      parent=styles["BodyText"],
      fontSize=10,
      leading=14,
      textColor=colors.HexColor("#2D3748"),
  )

  elements = []

  # Title Header
  elements.append(
      Paragraph("Job Placement Automated Analysis Report", title_style)
  )
  elements.append(
      Paragraph(
          "Executive Summary & Data-Driven Performance Conclusion",
          subtitle_style,
      )
  )
  elements.append(
      HRFlowable(
          width="100%",
          thickness=1.5,
          color=colors.HexColor("#CBD5E0"),
          spaceAfter=15,
      )
  )

  # Section 1: Executive Summary Metrics
  elements.append(Paragraph("1. Executive Key Performance Metrics", h2_style))

  metrics_data = [
      ["Metric", "Value", "Metric", "Value"],
      ["Total Candidates", f"{total_candidates:,}", "Placed Count", f"{placed_count:,}"],
      [
          "Not Placed Count",
          f"{not_placed_count:,}",
          "Placement Rate",
          f"{placement_rate:.2f}%",
      ],
      [
          "Avg Technical Score",
          f"{avg_tech:.2f}",
          "Avg Skills Match",
          f"{avg_skill:.2f}%",
      ],
  ]

  table = Table(metrics_data, colWidths=[130, 120, 130, 120])
  table.setStyle(
      TableStyle([
          ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#2B6CB0")),
          ("TEXTCOLOR", (0, 0), (-1, 0), colors.whitesmoke),
          ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
          ("ALIGN", (0, 0), (-1, -1), "CENTER"),
          (
              "ROWBACKGROUNDS",
              (0, 1),
              (-1, -1),
              [colors.HexColor("#F7FAFC"), colors.white],
          ),
          ("GRID", (0, 0), (-1, -1), 0.5, colors.HexColor("#E2E8F0")),
          ("BOTTOMPADDING", (0, 0), (-1, -1), 6),
          ("TOPPADDING", (0, 0), (-1, -1), 6),
      ])
  )
  elements.append(table)
  elements.append(Spacer(1, 15))

  # Section 2: Visual Distribution & Overview
  elements.append(Paragraph("2. Placement Distribution Analysis", h2_style))

  summary_p = Paragraph(
      f"The overall evaluation shows a <b>{placement_rate:.2f}%</b> placement"
      f" rate across {total_candidates:,} evaluated profiles. Candidates"
      f" averaged a technical score of <b>{avg_tech:.2f}</b> and an average"
      f" experience of <b>{avg_exp:.2f} years</b>.",
      body_style,
  )

  img = Image(pie_chart_path, width=220, height=165)
  layout_table = Table([[summary_p, img]], colWidths=[300, 200])
  layout_table.setStyle(
      TableStyle([
          ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
          ("ALIGN", (1, 0), (1, 0), "RIGHT"),
      ])
  )
  elements.append(layout_table)
  elements.append(Spacer(1, 15))

  # Section 3: Automated Analytical Conclusion
  elements.append(Paragraph("3. Overall Conclusion & Recommendations", h2_style))

  conclusion_text = f"""
    <b>Analytical Verdict:</b> {performance_verdict}<br/><br/>
    <b>Key Takeaways:</b><br/>
    • <b>Placement Ratio:</b> Out of {total_candidates:,} candidates, {placed_count:,} secured offers, resulting in a non-placement pool of {not_placed_count:,}.<br/>
    • <b>Technical Alignment:</b> Candidate technical capabilities strongly correlate with final placement results.<br/><br/>
    <b>Strategic Recommendation:</b> {rec_text}
    """
  elements.append(Paragraph(conclusion_text, body_style))

  # Build Document
  doc.build(elements)

  # Clean up temporary chart image
  if os.path.exists(pie_chart_path):
    os.remove(pie_chart_path)

  print(f"Report successfully generated: {output_pdf}")


# =====================================================
# EXAMPLE RUN WITH SYNTHETIC DATA
# =====================================================
if __name__ == "__main__":
  np.random.seed(42)
  n = 500
  sample_df = pd.DataFrame({
      "status": np.random.choice(
          ["Placed", "Not Placed"], size=n, p=[0.65, 0.35]
      ),
      "technical_score": np.random.normal(75, 10, size=n),
      "skills_match_percentage": np.random.normal(80, 8, size=n),
      "years_of_experience": np.random.exponential(2, size=n),
  })

  # Generate PDF Report
  generate_automated_report(sample_df)
```
